### TPC3: Corrida para o 100

O total começa em 0. O jogador e o computador alternam somando um número de 1 a 10 ao total. Quem atingir exatamente o número 100 vence.
- Se o computador jogar primeiro deverá vencer;
- Se o computador jogar em segundo poderá vencer se o outro jogador fizer uma aposta fora da estratégia vencedora.

Implementa o jogo com 2 vertentes: o computador joga primeiro (deverá ganhar sempre), e o computador joga em segundo lugar (poderá ganhar ou não dependendo das jogadas do outro).

```python
def jogo_computador_primeiro():
    print("\n--- Computador Joga Primeiro ---")
    total = 0
    jogada_comp = 1
    total = total + jogada_comp
    print(f"O computador jogou: {jogada_comp}. Total atual: {total}")
    
    ativo = True
    while ativo:
        jogada_jogador = 0
        while jogada_jogador < 1 or jogada_jogador > 10:
            jogada_jogador = int(input("Escolhe um número de 1 a 10: "))
            if jogada_jogador < 1 or jogada_jogador > 10:
                print("Número inválido. Tem de ser entre 1 e 10.")
        
        if total + jogada_jogador > 100:
            print("Número inválido, uma vez que o total excede o número 100!")
        else:
            total = total + jogada_jogador
            print(f"O jogador jogou: {jogada_jogador}. Total atual: {total}")
            
            # Verificar imediatamente se o jogador ganhou ANTES do computador jogar
            if total == 100:
                print("Parabéns! Venceste o jogo.")
                ativo = False
            else:
                jogada_comp = 11 - jogada_jogador
                total = total + jogada_comp
                print(f"O computador jogou: {jogada_comp}. Total atual: {total}")
                
                if total == 100:
                    print("O computador atingiu 100 e venceu!")
                    ativo = False

def jogo_computador_segundo():
    print("\n--- Computador Joga em Segundo ---")
    total = 0
    ativo = True
    posicoes_vencedoras = [1, 12, 23, 34, 45, 56, 67, 78, 89, 100]
    
    while ativo:
        jogada_jogador = 0
        while jogada_jogador < 1 or jogada_jogador > 10:
            jogada_jogador = int(input("Escolhe um número de 1 a 10: "))
            if jogada_jogador < 1 or jogada_jogador > 10:
                print("Número inválido. Tem de ser entre 1 e 10.")
        
        if total + jogada_jogador > 100:
            print("Número inválido, uma vez que o total excede o número 100!")
        else:
            total = total + jogada_jogador
            print(f"O jogador jogou: {jogada_jogador}. Total atual: {total}")
            
            # Verificar imediatamente se o jogador ganhou ANTES do computador jogar
            if total == 100:
                print("Parabéns! Atingiste 100 e venceste!")
                ativo = False
            else:
                jogada_comp = 1
                encontrou = False
                
                i = 0
                while i < len(posicoes_vencedoras):
                    alvo = posicoes_vencedoras[i]
                    if alvo > total and alvo - total <= 10 and not encontrou:
                        jogada_comp = alvo - total
                        encontrou = True
                    i = i + 1
                
                total = total + jogada_comp
                print(f"O computador jogou: {jogada_comp}. Total atual: {total}")
                
                if total == 100:
                    print("O computador atingiu 100 e venceu!")
                    ativo = False

def main():
    print("Bem-vindo ao Jogo do 100!")
    
    menu_ativo = True
    while menu_ativo:
        print("1. Computador joga primeiro")
        print("2. Computador joga em segundo")
        print("3. Sair")
        
        opcao = int(input("Opção (1/2/3): "))
        if opcao == 1:
            jogo_computador_primeiro()
        elif opcao == 2:
            jogo_computador_segundo()
        elif opcao == 3:
            print("Até à próxima!")
            menu_ativo = False
        else:
            print("Opção inválida.")

if __name__ == "__main__":
    main()
```
