```python
def criarTurma():
    turma=[]
    return turma

def inserirAluno(turma):
    nome = input("Nome do aluno: ")
    id = input("ID do aluno: ")
    notaTPC = (input("Nota do TPC: "))
    notaProj = (input("Nota do Projeto: "))
    notaTeste = (input("Nota do Teste: "))
    notas = [notaTPC, notaProj, notaTeste]
    aluno = (nome, id, notas)
    turma.append(aluno)
    print("Aluno inserido à turma.")
    return turma

def listarTurma(turma):
    print("\n Lista de alunos:{turma}")
    for aluno in turma:
        print(f"Nome: {aluno[0]}, ID: {aluno[1]}, Notas: {aluno[2]}")
    if not turma:
        print("Turma vazia.")
    return

def consultarAluno(turma, i):
    id = input("ID do aluno a consultar: ")
    encontrado = False
    for aluno in turma:
        if aluno[1] == id:
            print(f" Nome: {aluno[0]}, Notas: {aluno[2]}")
            encontrado = True
    if not encontrado:
        print("Aluno não encontrado.")
    return

def guardarTurma(turma, fnome):
    file = open(fnome,"w")
    res = ""
    for aluno in turma:
        nome, id, notas = aluno
        res += str(nome) + ";" + str(id) + ";" + str(notas[0]) + "::" + str (notas[1]) + "::" + str(notas[2]) + "|"
        res += "\n"
    file.write(res)
    file.close()

def recuperarTurma(fnome):
    turma = []
    file = open(fnome,"r")
    text = file.read()
    turma_text = text.split("\n")

    for p_text in turma_text[:-1]:
        t = []
        t_text=p_text.split("|")

        for t_text in t_text[:-1]:
            nome, id, notas = t_text.split(";")

            notas_lista=[]
            for n in notas.split("::"):
                notas_lista.append((n))
            aluno=(str(nome),int(id),list(notas_lista))
            t.append(aluno)
        turma.append(t)
    file.close()
    return turma

menu = -1
print ("""
📚 Menu de Gestão da Turma
1 - Criar uma turma
2 - Inserir um aluno
3 - Listar a turma
4 - Consultar aluno por ID
5 - Guardar turma em ficheiro
6 - Carregar turma de ficheiro
0 - Sair
        """)


while menu != 6:
    menu = int(input("O que deseja efetuar?"))
    if menu == 0:
        print("Terminou")
    elif menu == 1:
        turma=criarTurma()
    elif menu == 2:
        inserirAluno(turma)
    elif menu == 3:
        listarTurma(turma)
    elif menu == 4:
        consultarAluno(turma,id)
    elif menu == 5:
        print(f"A sua turma foi guardada no ficheiro: turma")
        guardarTurma(turma,"turma.txt")
    elif menu == 6:
        print(recuperarTurma("turma.txt"))
    else:
        print("Tente outro número")
```
